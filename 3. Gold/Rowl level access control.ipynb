# **Row Level Access Control**
users in United Airlines group will only be able to access the record with IATA_CODE = 'UA'

**Note** - I have created RLAC for United airlines only for this project but this can be implemented for all the airlines in case of real world project



```sql
select is_account_group_memeber('account user')
```

```sql
%sql
SHOW GROUPS
```

```sql
%sql
CREATE GROUP united_airlines_group;
```

```sql
%sql
ALTER GROUP united_airlines_group ADD USER `lunatuna004tunes_gmail.com#ext#@lunatuna004tunesgmail.onmicrosoft.com`;
```

```sql
%sql
CREATE OR REPLACE FUNCTION aviation_project.gold.united_airlines_filter(
  IATA_CODE STRING
)
RETURN
  is_member('united_airlines_group') AND IATA_CODE = 'UA';
```

```sql
%sql
ALTER TABLE aviation_project.gold.dim_airlines
SET ROW FILTER aviation_project.gold.united_airlines_filter
ON (IATA_CODE);
```

```sql
REVOKE ALL PRIVILEGES
ON TABLE aviation_catalog.gold.dim_airlines
FROM `united_airlines_group`;
```

```sql
ALTER TABLE aviation_project.gold.dim_airlines
DROP ROW FILTER;
```

```sql
DROP GROUP united_airlines_group;
```
